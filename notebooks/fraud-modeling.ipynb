```python
import pandas as pd

X_train = pd.read_csv(
    "../data/processed/X_train.csv"
)

X_test = pd.read_csv(
    "../data/processed/X_test.csv"
)

y_train = pd.read_csv(
    "../data/processed/y_train.csv"
).squeeze()

y_test = pd.read_csv(
    "../data/processed/y_test.csv"
).squeeze()

import sys
import os
sys.path.append(os.path.abspath('../'))
from src.evaluation import evaluate_model
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr.fit(
    X_train,
    y_train
)
```

```python
from src.evaluation import evaluate_model

lr_results, lr_cm = evaluate_model(
    lr,
    X_test,
    y_test
)

lr_results
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(
    lr_cm,
    annot=True,
    fmt="d"
)

plt.title(
    "Logistic Regression Confusion Matrix"
)

plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)
```

```python
rf_results, rf_cm = evaluate_model(
    rf,
    X_test,
    y_test
)

rf_results
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [10, 20],
    "min_samples_split": [2, 5]
}

grid = GridSearchCV(
    RandomForestClassifier(
        random_state=42
    ),
    param_grid,
    scoring="f1",
    cv=3,
    n_jobs=-1
)

grid.fit(
    X_train,
    y_train
)
```

```python
best_rf = grid.best_estimator_
```

```python
from sklearn.model_selection import (
    StratifiedKFold,
    cross_val_score
)

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
lr_f1 = cross_val_score(
    lr,
    X_train,
    y_train,
    scoring="f1",
    cv=skf
)
```

```python
rf_f1 = cross_val_score(
    best_rf,
    X_train,
    y_train,
    scoring="f1",
    cv=skf
)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],
    "AUC-PR": [
        lr_results["AUC_PR"],
        rf_results["AUC_PR"]
    ],
    "F1-Score": [
        lr_results["F1_Score"],
        rf_results["F1_Score"]
    ]
})

comparison
```

Random Forest was selected as the final model because it achieved higher AUC-PR and F1-Score than Logistic Regression while maintaining stable cross-validation performance. Although Logistic Regression offers greater interpretability, the performance improvements obtained from Random Forest are valuable in fraud detection where missed fraudulent transactions can result in significant financial losses.

```python
from sklearn.model_selection import cross_val_score

lr_auc = cross_val_score(
    lr,
    X_train,
    y_train,
    scoring="average_precision",
    cv=skf,
    n_jobs=-1
)

rf_auc = cross_val_score(
    rf,
    X_train,
    y_train,
    scoring="average_precision",
    cv=skf,
    n_jobs=-1
)
```

```python
lr_f1.mean(), lr_f1.std()

rf_f1.mean(), rf_f1.std()
```

```python
lr_auc.mean(), lr_auc.std()
rf_auc.mean(), rf_auc.std()
```

```python
print("Logistic Regression AUC-PR")
print("Mean:", lr_auc.mean())
print("Std:", lr_auc.std())

print("\nRandom Forest AUC-PR")
print("Mean:", rf_auc.mean())
print("Std:", rf_auc.std())
```

```python
print("Logistic Regression F1")
print("Mean:", lr_f1.mean())
print("Std:", lr_f1.std())

print("\nRandom Forest F1")
print("Mean:", rf_f1.mean())
print("Std:", rf_f1.std())
```

```python
cv_results = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "CV F1 Mean": [lr_f1.mean(), rf_f1.mean()],
    "CV F1 Std": [lr_f1.std(), rf_f1.std()],
    "CV AUC-PR Mean": [lr_auc.mean(), rf_auc.mean()],
    "CV AUC-PR Std": [lr_auc.std(), rf_auc.std()]
})

cv_results
```
